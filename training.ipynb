```python
from gymnasium.wrappers import FlattenObservation
from sympy.physics.units import action

from snakeagent import SnakeAgent
from snakeenv import SnakeEnv

from tqdm import tqdm
import torch
```

```python
torch.cuda.is_available()
```

```python
env = FlattenObservation(SnakeEnv(render_mode=None, size_x=10, size_y=10))
agent = SnakeAgent(
    env=env,
    batch_size=5,
    gamma=0.99,
    epsilon_start=1.0,
    epsilon_end=0.01,
    epsilon_decay=0.995,
    target_net_update_rate=100,
    optimizer_learning_rate=0.001
)

print(env.observation_space)
print(env.action_space)

n_episodes = 15
for episode in tqdm(range(n_episodes)):
    observation, info = env.reset()
    done = False

    while not done:
        action = agent.get_action(observation)

        next_observation, reward, terminated, truncated, info = env.step(int(action))

        agent.update(observation, action, reward, terminated, next_observation)

        done = terminated or truncated
        observation = next_observation
env.close()
```

```python
env = FlattenObservation(SnakeEnv(render_mode="human", size_x=10, size_y=10))
observation, info = env.reset()
done = False
while not done:
    action = agent.get_action(observation, exploit_only=True)
    next_observation, reward, terminated, truncated, info = env.step(int(action))
    done = terminated or truncated
    observation = next_observation
env.close()
```
