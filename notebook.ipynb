First, we prepare the agent:

```python
import torch
from snakeagent import SnakeAgent

# Automatically use GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Local path for saving the model (can be changed to keep multiple versions):
save_path = "model.pth"

# Create the agent (hyperparameters can be adjusted if needed):
agent = SnakeAgent(
    device=device,
    game_size_x=5,                  # must match between training and playing
    game_size_y=5,                  # must match between training and playing
    num_episodes=4000,              # number of games to play during training
    gamma=0.95,                     # discount factor (0–1), how much future rewards are worth
    epsilon_start=1.0,              # initial exploration rate
    epsilon_end=0.05,               # final exploration rate
    epsilon_end_episode = 1000,     # number of episodes over which epsilon decays to final value
    learning_rate=0.0003,           # learning rate for Adam optimizer
    target_net_update_rate=0.005,   # soft update rate for target network
    batch_size=512,                 # minibatch size for training
    replay_memory_size=10000        # maximum capacity of replay memory
)
```

Now, train the model using the agent:

```python
agent.train(save_path)
```

Use the trained model (with the provided save path) to display the game window and watch it play.

```python
# A pretrained example is included in the repository. To use it, set save_path to `model_example.pth`.
agent.play_once(save_path)
```

(Optional) For PixiJS exploration, export the model in ONNX format.

```python
agent.export_onnx(save_path, "model.onnx")
```
